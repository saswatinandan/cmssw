```python
hist_sig = file_['sep19_2_2_dump_raw/adc_idx_bkg']
hist_values = hist_sig.values()   # 2D numpy array of bin contents
x_edges, y_edges = hist_sig.axes[0].edges(), hist_sig.axes[1].edges()  # Lists of bin edges: [x_edges, y_edges]

plt.figure(figsize=(7,6))
plt.pcolormesh(x_edges[:11], y_edges, (hist_values[:10,]/np.sum(hist_values)).T, shading='auto', cmap='viridis')
plt.xlabel("strip idx")
plt.ylabel("ADC value")
plt.yticks(np.arange(0, 260, 10))
plt.title("strip idx (upto 10) vs ADC value for non-associated cluster")
plt.colorbar(label="Normalized Bin content")
plt.show()
```

```python
def data_loader(file_path, var, tree, signal, frac=0):
    print('calling')
    with uproot.open(file)[f'sep19_2_2_dump_raw/{tree}'] as input_tree:
         df = input_tree.arrays(var)
         df = df[(df['target']==signal)]
         df = pd.DataFrame({k: ak.to_list(df[k]) for k in df.fields})
         if frac:
             df = df.sample(frac=frac, random_state=42)
         return df
```

```python
def plot_var_dist(df, varlists):
   for var in varlists:
       if var == 'target': continue
       print(var)
       minimum_value = min(df[var])
       maximum_value = max(df[var])
       bins = np.linspace(minimum_value, maximum_value, 100)
       for label in [0,1]:
           data = df[(df['target'] == label)][var]
           weights = np.ones_like(data) / len(data)
           plt.hist(data, weights=weights, histtype='step', \
            label='sig' if label==1 else 'bkg', bins=bins)
       plt.legend(loc='best', title=var)
       plt.ylim([0.0001, 10])
       plt.yscale('log')
       plt.savefig(f'{var}.png')
       plt.show()
```

```python
vars = ['target',
        'size', 'charge', 'adcs_ten',\
            'mean_x', 'mean_y', 'mean_z', 'eta', 'phi']            
print('loading sig')
sig = data_loader(file, vars, 'tree', 1, 1)
print('loading bg')
bkg = data_loader(file, vars, 'tree', 0, 1)
df = pd.concat([sig, bkg], ignore_index=True)
df[[f'adc_at_idx_{i}' for i in range(4)]] = pd.DataFrame(df['adcs_ten'].apply(lambda x: x[:4]).tolist(), index=df.index)
vars.extend([f'adc_at_idx_{i}' for i in range(4)])
vars.remove('adcs_ten')
df.drop(columns=['adcs_ten'])
```

```python
print(vars)
plot_var_dist(df, vars)
print('plotting done')
```
